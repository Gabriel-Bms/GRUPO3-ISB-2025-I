## 1. REPOSO

```python
fs = 1000

signal1 = OpenSignalsReader('Señales ECG Raw/2 ecg_reposo_1.1.txt')
signal2 = OpenSignalsReader('Señales ECG Raw/3 ecg_reposo_1.2.txt')
signal3 = OpenSignalsReader('Señales ECG Raw/4 ecg_reposo_2.1.txt')
signal4 = OpenSignalsReader('Señales ECG Raw/5 ecg_reposo_2.2.txt')
signal5 = OpenSignalsReader('Señales ECG Raw/6 ecg_reposo_3.1.txt')
signal6 = OpenSignalsReader('Señales ECG Raw/7 ecg_reposo_3.2.txt')


fig, axs = plt.subplots(6, 1, figsize=(15, 12), sharex=True)

# Lista de señales
signals = [signal1, signal2, signal3, signal4, signal5, signal6]

# Graficar en un bucle
for i, sig in enumerate(signals):
    axs[i].plot(np.arange(len(sig.signal(2))) / fs, sig.signal(2))
    axs[i].set_ylabel(f'Señal {i+1}')
    axs[i].set_xlim(0, 30)

# Configurar el último eje
axs[-1].set_xlabel('Tiempo (s)')

plt.grid(True)
plt.show()
```
