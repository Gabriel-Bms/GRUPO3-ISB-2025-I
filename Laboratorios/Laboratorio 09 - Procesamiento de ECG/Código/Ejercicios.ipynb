```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis
from sklearn.decomposition import PCA
import neurokit2 as nk


# 1) GENERAR TRES SEÑALES DE ECG SINTÉTICAS
fs = 500  # frecuencia de muestreo (Hz)
duration = 10  # segundos

np.random.seed(42)  #

# (a) ECG “normal” (una sola derivación) mediante ecgsyn
ti0 = np.array((-70, -15, 0, 15, 100))
ai0 = np.array(( 1.2, -5, 30, -7.5, 0.75))
bi0 = np.array((0.25, 0.1,  0.1,  0.1,  0.4))

ecg_single = nk.ecg_simulate(
    duration=duration,
    sampling_rate=fs,
    heart_rate=70,
    method="ecgsyn",
    ti = ti0,
    ai = ai0,
    bi = bi0
)

# (b) ECG “modificado” (ligera variación de amplitud y ancho)
ti1 = np.random.normal(ti0, np.ones(5) * 3)
ai1 = np.random.normal(ai0, np.abs(ai0/5))
bi1 = np.random.normal(bi0, np.abs(bi0/5))

ecg_mod = nk.ecg_simulate(
    duration=duration,
    sampling_rate=fs,
    heart_rate=70,
    method="ecgsyn",
    ti = ti1,
    ai = ai1,
    bi = bi1
)

# (c) ECG “anómalo” (QRS ancho + T invertida)
ti2 = np.random.normal(ti0, np.ones(5) * 3)
ai2 = np.random.normal(ai0, np.abs(ai0/5))
bi2 = np.random.normal(bi0, np.abs(bi0/5))
ai2[4] = -0.5

bi2[2] = bi2[2] * 2

ecg_abnormal = nk.ecg_simulate(
    duration=duration,
    sampling_rate=fs,
    heart_rate=70,
    method="ecgsyn",
    ti = ti2,
    ai = ai2,
    bi = bi2
)

ecg_signals = [ecg_single, ecg_mod, ecg_abnormal]
signal_names = ['ECG_Single', 'ECG_Mod', 'ECG_Abnormal']

# 2) EXTRAER CARACTERÍSTICAS DE CADA SEÑAL & ARMAR UN DF
all_features = []

for i, signal in enumerate(ecg_signals):
    # 2.1) Detectar picos R (para obtener RR)
    ecg_cleaned = nk.ecg_clean(signal, sampling_rate=fs, method="biosppy")
    peaks_dict = nk.ecg_peaks(ecg_cleaned, sampling_rate=fs)
    rpeaks = peaks_dict[0]

    # 2.2) Calcular RR intervals en segundos
    if len(rpeaks) >= 2:
        t_peaks = rpeaks / fs
        rr_intervals = np.diff(t_peaks)
        rr_mean = np.mean(rr_intervals)
        rr_std  = np.std(rr_intervals)
    else:
        rr_mean = np.nan
        rr_std  = np.nan

    # 2.3) Estadísticos “básicos” + HOS
    feats = {
        'Signal_Type'       : signal_names[i],
        'Mean'              : np.mean(signal),
        'Median'            : np.median(signal),
        'STD'               : np.std(signal),
        'Skewness'          : skew(signal),
        'Kurtosis'          : kurtosis(signal, fisher=True),  
    }

    all_features.append(pd.DataFrame([feats]))

combined_features_df = pd.concat(all_features, ignore_index=True)

# 3) HACER PCA DE 2 COMPONENTES SOBRE ESAS FEATURES
pca_input = combined_features_df.drop(columns=['Signal_Type']).copy()

pca_input = pca_input.fillna(pca_input.mean())

pca = PCA(n_components=2)
pca_result = pca.fit_transform(pca_input)

pca_df = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])
pca_df['Signal_Type'] = combined_features_df['Signal_Type']

# 4) GRAFICAR RESULTADOS DEL PCA
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x='PC1',
    y='PC2',
    hue='Signal_Type',
    data=pca_df,
    s=100,
    palette="Set2"
)
plt.title('PCA de Características de ECG (2 Dimensiones)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.grid(alpha=0.3)
plt.legend(title='Tipo de Señal')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis
from sklearn.decomposition import PCA
import neurokit2 as nk
import pywt  
# 1) GENERAR SEÑALES ADICIONALES (basadas en las tres originales)

fs = 500  
duration = 10 

np.random.seed(42)  

ti0 = np.array((-70, -15, 0, 15, 100))
ai0 = np.array(( 1.2, -5, 30, -7.5, 0.75))
bi0 = np.array((0.25, 0.1,  0.1,  0.1,  0.4))

ecg_single_2 = nk.ecg_simulate(duration=duration, sampling_rate=fs, heart_rate=70, method="ecgsyn", ti=ti0, ai=ai0, bi=bi0)
ecg_mod_2 = nk.ecg_simulate(duration=duration, sampling_rate=fs, heart_rate=70, method="ecgsyn", ti=np.random.normal(ti0, np.ones(5) * 3), ai=np.random.normal(ai0, np.abs(ai0 / 5)), bi=np.random.normal(bi0, np.abs(bi0 / 5)))
ecg_abnormal_2 = nk.ecg_simulate(duration=duration, sampling_rate=fs, heart_rate=70, method="ecgsyn", ti=np.random.normal(ti0, np.ones(5) * 3), ai=np.random.normal(ai0, np.abs(ai0 / 5)), bi=np.random.normal(bi0, np.abs(bi0 / 5)))

ecg_signals = [ecg_single, ecg_mod, ecg_abnormal, ecg_single_2, ecg_mod_2, ecg_abnormal_2]
signal_names = ['ECG_Single', 'ECG_Mod', 'ECG_Abnormal', 'ECG_Single_2', 'ECG_Mod_2', 'ECG_Abnormal_2']

# 2) EXTRAER CARACTERÍSTICAS DE CADA SEÑAL Y ARMAR UN DF
all_features = []

for i, signal in enumerate(ecg_signals):
    ecg_cleaned = nk.ecg_clean(signal, sampling_rate=fs, method="biosppy")
    peaks_dict = nk.ecg_peaks(ecg_cleaned, sampling_rate=fs)
    rpeaks = peaks_dict[0]

    if len(rpeaks) >= 2:
        t_peaks = rpeaks / fs
        rr_intervals = np.diff(t_peaks)
        rr_mean = np.mean(rr_intervals)
        rr_std  = np.std(rr_intervals)
    else:
        rr_mean = np.nan
        rr_std  = np.nan

    feats = {
        'Signal_Type': signal_names[i],
        'Mean': np.mean(signal),
        'Median': np.median(signal),
        'STD': np.std(signal),
        'Skewness': skew(signal),
        'Kurtosis': kurtosis(signal, fisher=True),
    }

    coeffs = pywt.dwt(signal, 'db4') 
    cA, cD = coeffs 

    print(f"Coeficientes DWT para la señal {signal_names[i]}:")
    print("Coeficientes de Aproximación:", cA[:10]) 
    print("Coeficientes de Detalle:", cD[:10])  
    print("-" * 40)

    feats['DWT_Coef_Approx'] = np.mean(cA) 
    feats['DWT_Coef_Detail'] = np.mean(cD)  

    all_features.append(pd.DataFrame([feats]))

combined_features_df = pd.concat(all_features, ignore_index=True)

# 3) REDUCCIÓN DE DIMENSIONALIDAD A 2 COMPONENTES (PCA)
pca_input = combined_features_df.drop(columns=['Signal_Type']).copy()

pca_input = pca_input.fillna(pca_input.mean())

pca = PCA(n_components=2)
pca_result = pca.fit_transform(pca_input)

pca_df = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])
pca_df['Signal_Type'] = combined_features_df['Signal_Type']

# 4) GRAFICAR RESULTADOS DEL PCA
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x='PC1',
    y='PC2',
    hue='Signal_Type',
    data=pca_df,
    s=100,
    palette="Set2"
)
plt.title('PCA de Características de ECG (2 Dimensiones)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.grid(alpha=0.3)
plt.legend(title='Tipo de Señal')
plt.tight_layout()
plt.show()
```
